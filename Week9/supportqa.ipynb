```python
# ============================================================
# SUPPORTIQ — CELL 03 (FIXED WITH ALL IMPORTS)
# ============================================================

from pathlib import Path
from urllib.parse import urlparse
import hashlib
import json
import re
import shutil
import uuid
import html
import time
import math
from datetime import datetime, timezone
from rank_bm25 import BM25Okapi  # <-- ADDED THIS

print("=" * 60)
print("SUPPORTIQ CELL 03 - ALL FUNCTIONS")
print("=" * 60)

# ------------------------------------------------------------
# 1. FILE PATHS
# ------------------------------------------------------------

DOCUMENTS_JSON = DATA_DIR / "documents.json"
CHUNKS_JSON = DATA_DIR / "chunks.json"
EMBEDDINGS_PATH = INDEX_DIR / "embeddings.npy"
METADATA_PATH = INDEX_DIR / "chunk_metadata.json"
INDEX_INFO_PATH = INDEX_DIR / "index_info.json"
TICKETS_JSON = TICKET_DIR / "tickets.json"

# ------------------------------------------------------------
# 2. TEXT CLEANING
# ------------------------------------------------------------

def clean_text(text):
    if not text:
        return ""
    text = html.unescape(str(text))
    text = text.replace("\r\n", "\n").replace("\r", "\n")
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n{3,}", "\n\n", text)
    return text.strip()

def estimate_tokens(text):
    return len(re.findall(r"\S+", text))

def split_sentences(text):
    text = clean_text(text)
    if not text:
        return []
    abbr = {"e.g.": "eg__", "i.e.": "ie__", "Mr.": "Mr__", "Mrs.": "Mrs__", "Dr.": "Dr__"}
    for old, new in abbr.items():
        text = text.replace(old, new)
    sentences = re.split(r"(?<=[.!?])\s+|\n{2,}", text)
    result = []
    for s in sentences:
        s = s.strip()
        if s:
            for old, new in abbr.items():
                s = s.replace(new, old)
            result.append(s)
    return result

# ------------------------------------------------------------
# 3. CHUNKING
# ------------------------------------------------------------

def chunk_text(text):
    sentences = split_sentences(text)
    if not sentences:
        return []
    chunks = []
    current = []
    current_tokens = 0
    for sentence in sentences:
        sent_tokens = estimate_tokens(sentence)
        if sent_tokens > CHUNK_SIZE:
            if current:
                chunks.append(" ".join(current))
                current = []
                current_tokens = 0
            tokens = re.findall(r"\S+", sentence)
            start = 0
            while start < len(tokens):
                end = min(start + CHUNK_SIZE, len(tokens))
                piece = " ".join(tokens[start:end])
                if piece.strip():
                    chunks.append(piece)
                start = end - CHUNK_OVERLAP if end - CHUNK_OVERLAP > start else end
            continue
        if current_tokens + sent_tokens <= CHUNK_SIZE:
            current.append(sentence)
            current_tokens += sent_tokens
        else:
            if current:
                chunks.append(" ".join(current))
            overlap_sentences = []
            overlap_tokens = 0
            for prev in reversed(current):
                prev_tokens = estimate_tokens(prev)
                if overlap_tokens + prev_tokens > CHUNK_OVERLAP:
                    break
                overlap_sentences.insert(0, prev)
                overlap_tokens += prev_tokens
            current = overlap_sentences + [sentence]
            current_tokens = overlap_tokens + sent_tokens
    if current:
        chunks.append(" ".join(current))
    return [c for c in chunks if len(c) >= MIN_CHUNK_CHARACTERS]

# ------------------------------------------------------------
# 4. FILE EXTRACTION
# ------------------------------------------------------------

from pypdf import PdfReader
from docx import Document

def extract_pdf(file_path):
    pages = []
    reader = PdfReader(str(file_path))
    for i, page in enumerate(reader.pages, 1):
        try:
            text = clean_text(page.extract_text() or "")
            if text:
                pages.append({"page": i, "text": text})
        except:
            continue
    return "\n\n".join(p["text"] for p in pages), pages

def extract_docx(file_path):
    doc = Document(str(file_path))
    sections = []
    for para in doc.paragraphs:
        text = clean_text(para.text)
        if text:
            sections.append(text)
    for table in doc.tables:
        for row in table.rows:
            cells = [clean_text(c.text) for c in row.cells if clean_text(c.text)]
            if cells:
                sections.append(" | ".join(cells))
    return "\n\n".join(sections), []

def extract_txt(file_path):
    for enc in ["utf-8", "utf-8-sig", "cp1252"]:
        try:
            with open(file_path, "r", encoding=enc) as f:
                return clean_text(f.read()), []
        except:
            continue
    with open(file_path, "r", encoding="utf-8", errors="replace") as f:
        return clean_text(f.read()), []

def extract_file(file_path):
    ext = file_path.suffix.lower()
    if ext == ".pdf":
        return extract_pdf(file_path)
    elif ext == ".docx":
        return extract_docx(file_path)
    elif ext == ".txt":
        return extract_txt(file_path)
    raise ValueError(f"Unsupported: {ext}")

# ------------------------------------------------------------
# 5. FILE HASH
# ------------------------------------------------------------

def calculate_sha256(file_path):
    sha = hashlib.sha256()
    with open(file_path, "rb") as f:
        for block in iter(lambda: f.read(65536), b""):
            sha.update(block)
    return sha.hexdigest()

# ------------------------------------------------------------
# 6. DOCUMENT OPERATIONS
# ------------------------------------------------------------

def load_documents():
    if DOCUMENTS_JSON.exists():
        with open(DOCUMENTS_JSON, "r", encoding="utf-8") as f:
            data = json.load(f)
            return data if isinstance(data, list) else []
    return []

def save_documents(docs):
    with open(DOCUMENTS_JSON, "w", encoding="utf-8") as f:
        json.dump(docs, f, indent=2, ensure_ascii=False)

# ------------------------------------------------------------
# 7. INGEST LOCAL FILE
# ------------------------------------------------------------

def ingest_local_file(file_path, org_id=DEFAULT_ORGANIZATION_ID):
    print(f"\nIngesting: {file_path}")
    source = Path(str(file_path)).resolve()
    if not source.exists():
        raise FileNotFoundError(f"File not found: {source}")
    
    ext = source.suffix.lower()
    if ext not in [".pdf", ".docx", ".txt"]:
        raise ValueError(f"Unsupported: {ext}")
    
    size_mb = source.stat().st_size / (1024 * 1024)
    if size_mb > MAX_DOCUMENT_SIZE_MB:
        raise ValueError(f"File exceeds {MAX_DOCUMENT_SIZE_MB}MB")
    
    file_hash = calculate_sha256(source)
    
    docs = load_documents()
    for doc in docs:
        if doc.get("file_hash") == file_hash:
            print("Document already exists")
            return doc
    
    safe_name = re.sub(r"[^a-zA-Z0-9._-]+", "_", source.name)
    dest = DOCUMENT_DIR / f"{file_hash[:12]}_{safe_name}"
    shutil.copy2(source, dest)
    
    text, pages = extract_file(dest)
    if not text.strip():
        raise ValueError("No readable text extracted")
    
    doc_id = f"DOC-{uuid.uuid4().hex[:12].upper()}"
    record = {
        "document_id": doc_id,
        "organization_id": org_id,
        "source_type": "file",
        "source_name": source.name,
        "source_path": str(dest),
        "file_hash": file_hash,
        "text": text,
        "pages": pages,
        "character_count": len(text),
        "created_at": datetime.now(timezone.utc).isoformat(),
        "status": "READY"
    }
    
    docs.append(record)
    save_documents(docs)
    print(f"Document ingested: {doc_id}")
    return record

# ------------------------------------------------------------
# 8. INGEST WEB
# ------------------------------------------------------------

def ingest_web_url(url, org_id=DEFAULT_ORGANIZATION_ID):
    print(f"\nIngesting web: {url}")
    
    parsed = urlparse(url)
    if parsed.scheme.lower() not in ["http", "https"]:
        raise ValueError("Only HTTP/HTTPS supported")
    
    response = requests.get(url, timeout=20, headers={"User-Agent": "SupportIQ/2.0"})
    response.raise_for_status()
    
    soup = BeautifulSoup(response.text, "lxml")
    for tag in soup(["script", "style", "noscript", "svg", "nav", "footer", "header"]):
        tag.decompose()
    
    title = clean_text(soup.title.get_text(" ", strip=True)) if soup.title else url
    main = soup.find("main") or soup.find("article") or soup.body or soup
    text = clean_text(main.get_text("\n", strip=True))
    
    if not text:
        raise ValueError("No readable text found")
    
    content_hash = hashlib.sha256((url + "\n" + text).encode()).hexdigest()
    
    docs = load_documents()
    for doc in docs:
        if doc.get("file_hash") == content_hash:
            print("Web page already exists")
            return doc
    
    doc_id = f"WEB-{uuid.uuid4().hex[:12].upper()}"
    record = {
        "document_id": doc_id,
        "organization_id": org_id,
        "source_type": "web",
        "source_name": title,
        "source_url": response.url,
        "file_hash": content_hash,
        "text": text,
        "pages": [],
        "character_count": len(text),
        "created_at": datetime.now(timezone.utc).isoformat(),
        "status": "READY"
    }
    
    docs.append(record)
    save_documents(docs)
    print(f"Web page ingested: {doc_id}")
    return record

# ------------------------------------------------------------
# 9. BUILD CHUNKS
# ------------------------------------------------------------

def build_chunks(org_id=DEFAULT_ORGANIZATION_ID):
    print(f"\nBuilding chunks for: {org_id}")
    
    docs = load_documents()
    tenant_docs = [d for d in docs if d.get("organization_id") == org_id]
    
    if not tenant_docs:
        print("No documents found")
        save_chunks([])
        return []
    
    all_chunks = []
    for doc in tenant_docs:
        doc_id = doc.get("document_id")
        print(f"Processing: {doc.get('source_name')}")
        
        text_chunks = chunk_text(doc.get("text", ""))
        for i, chunk in enumerate(text_chunks, 1):
            chunk_id = f"{doc_id}-C{i:04d}"
            all_chunks.append({
                "chunk_id": chunk_id,
                "document_id": doc_id,
                "organization_id": org_id,
                "chunk_number": i,
                "text": chunk,
                "source_type": doc.get("source_type"),
                "source_name": doc.get("source_name"),
                "source_url": doc.get("source_url"),
                "page": None,
                "character_count": len(chunk),
                "token_estimate": estimate_tokens(chunk)
            })
        
        print(f"  Created {len(text_chunks)} chunks")
    
    save_chunks(all_chunks)
    print(f"Total chunks: {len(all_chunks)}")
    return all_chunks

def save_chunks(chunks):
    with open(CHUNKS_JSON, "w", encoding="utf-8") as f:
        json.dump(chunks, f, indent=2, ensure_ascii=False)

def load_chunks(org_id=None):
    if not CHUNKS_JSON.exists():
        return []
    with open(CHUNKS_JSON, "r", encoding="utf-8") as f:
        data = json.load(f)
    if org_id:
        return [c for c in data if c.get("organization_id") == org_id]
    return data

# ------------------------------------------------------------
# 10. BUILD VECTOR INDEX
# ------------------------------------------------------------

def build_vector_index(org_id=DEFAULT_ORGANIZATION_ID):
    print(f"\nBuilding vector index for: {org_id}")
    
    chunks = load_chunks(org_id)
    if not chunks:
        print("No chunks available")
        np.save(EMBEDDINGS_PATH, np.empty((0, EMBEDDING_DIMENSION), dtype=np.float32))
        with open(METADATA_PATH, "w", encoding="utf-8") as f:
            json.dump([], f, indent=2)
        return np.empty((0, EMBEDDING_DIMENSION), dtype=np.float32)
    
    texts = [c.get("text", "") for c in chunks]
    valid = [i for i, t in enumerate(texts) if t and t.strip()]
    chunks = [chunks[i] for i in valid]
    texts = [texts[i] for i in valid]
    
    print(f"Generating embeddings for {len(texts)} chunks...")
    embeddings = embedding_model.encode(
        texts,
        batch_size=EMBEDDING_BATCH_SIZE,
        show_progress_bar=True,
        convert_to_numpy=True,
        normalize_embeddings=True
    )
    embeddings = np.asarray(embeddings, dtype=np.float32)
    
    np.save(EMBEDDINGS_PATH, embeddings)
    
    metadata = []
    for chunk in chunks:
        metadata.append({
            "chunk_id": chunk.get("chunk_id"),
            "document_id": chunk.get("document_id"),
            "organization_id": chunk.get("organization_id"),
            "text": chunk.get("text"),
            "source_type": chunk.get("source_type"),
            "source_name": chunk.get("source_name"),
            "source_url": chunk.get("source_url"),
            "page": chunk.get("page"),
            "chunk_number": chunk.get("chunk_number"),
        })
    
    with open(METADATA_PATH, "w", encoding="utf-8") as f:
        json.dump(metadata, f, indent=2, ensure_ascii=False)
    
    print(f"Vector index built: {embeddings.shape[0]} vectors")
    return embeddings

# ------------------------------------------------------------
# 11. LOAD INDEX
# ------------------------------------------------------------

def load_vector_index():
    if not EMBEDDINGS_PATH.exists():
        return np.empty((0, EMBEDDING_DIMENSION), dtype=np.float32)
    return np.load(EMBEDDINGS_PATH)

def load_metadata():
    if not METADATA_PATH.exists():
        return []
    with open(METADATA_PATH, "r", encoding="utf-8") as f:
        return json.load(f)

def load_current_index():
    embeddings = load_vector_index()
    metadata = load_metadata()
    min_len = min(len(embeddings), len(metadata))
    return embeddings[:min_len], metadata[:min_len]

# ------------------------------------------------------------
# 12. BM25 TOKENS
# ------------------------------------------------------------

def bm25_tokens(text):
    return re.findall(r"\b[\w+#.-]+\b", str(text).lower())

# ------------------------------------------------------------
# 13. HYBRID SEARCH
# ------------------------------------------------------------

def reciprocal_rank_fusion(vector_results, keyword_results, k=60):
    scores = {}
    result_map = {}
    for rank, item in enumerate(vector_results, 1):
        chunk_id = item["chunk_id"]
        scores[chunk_id] = scores.get(chunk_id, 0.0) + 1.0 / (k + rank)
        result_map[chunk_id] = item
    for rank, item in enumerate(keyword_results, 1):
        chunk_id = item["chunk_id"]
        scores[chunk_id] = scores.get(chunk_id, 0.0) + 1.0 / (k + rank)
        result_map[chunk_id] = item
    fused = []
    for chunk_id, score in scores.items():
        item = dict(result_map[chunk_id])
        item["rrf_score"] = float(score)
        fused.append(item)
    fused.sort(key=lambda x: x["rrf_score"], reverse=True)
    return fused

def hybrid_search(query, org_id=DEFAULT_ORGANIZATION_ID):
    query = str(query).strip()
    if not query:
        return []
    
    embeddings, metadata = load_current_index()
    allowed = [i for i, item in enumerate(metadata) 
               if item.get("organization_id") == org_id]
    if not allowed:
        return []
    
    # Vector search
    q_emb = embedding_model.encode([query], convert_to_numpy=True, normalize_embeddings=True)[0]
    tenant_embeddings = embeddings[allowed]
    scores = tenant_embeddings @ q_emb
    ranking = np.argsort(scores)[::-1][:VECTOR_TOP_K]
    vector_results = []
    for rank_pos in ranking:
        original_idx = allowed[int(rank_pos)]
        item = dict(metadata[original_idx])
        item["vector_score"] = float(scores[rank_pos])
        vector_results.append(item)
    
    # BM25 search
    tenant_metadata = [metadata[i] for i in allowed]
    corpus = [bm25_tokens(item.get("text", "")) for item in tenant_metadata]
    query_tokens = bm25_tokens(query)
    if query_tokens:
        bm25 = BM25Okapi(corpus)
        bm25_scores = bm25.get_scores(query_tokens)
        ranking = np.argsort(bm25_scores)[::-1][:BM25_TOP_K]
        keyword_results = []
        for idx in ranking:
            idx = int(idx)
            item = dict(tenant_metadata[idx])
            item["keyword_score"] = float(bm25_scores[idx])
            keyword_results.append(item)
    else:
        keyword_results = []
    
    fused = reciprocal_rank_fusion(vector_results, keyword_results)
    return fused[:HYBRID_TOP_K]

# ------------------------------------------------------------
# 14. RERANK
# ------------------------------------------------------------

def sigmoid(value):
    value = float(value)
    if value >= 0:
        return 1.0 / (1.0 + math.exp(-value))
    return math.exp(value) / (1.0 + math.exp(value))

def retrieve_and_rerank(query, org_id=DEFAULT_ORGANIZATION_ID):
    candidates = hybrid_search(query, org_id)
    if not candidates:
        return []
    
    candidates = [c for c in candidates if c.get("organization_id") == org_id]
    if not candidates:
        return []
    
    pairs = [(query, c.get("text", "")) for c in candidates]
    raw_scores = reranker_model.predict(pairs, show_progress_bar=False)
    
    reranked = []
    for item, raw_score in zip(candidates, raw_scores):
        result = dict(item)
        raw_score = float(raw_score)
        result["reranker_raw_score"] = raw_score
        result["reranker_score"] = sigmoid(raw_score)
        reranked.append(result)
    
    reranked.sort(key=lambda x: x["reranker_raw_score"], reverse=True)
    reranked = reranked[:RERANK_TOP_K]
    
    return [r for r in reranked if r.get("reranker_score", 0) >= MIN_RERANK_SCORE]

# ------------------------------------------------------------
# 15. DETECT PROMPT INJECTION
# ------------------------------------------------------------

def detect_prompt_injection(text):
    text = str(text).lower()
    patterns = [
        r"ignore\s+(all\s+)?previous\s+instructions",
        r"disregard\s+(all\s+)?previous\s+instructions",
        r"system\s+prompt",
        r"reveal\s+(your\s+)?instructions",
        r"show\s+(me\s+)?your\s+prompt",
        r"forget\s+(all\s+)?previous",
    ]
    for pattern in patterns:
        if re.search(pattern, text, flags=re.IGNORECASE):
            return True
    return False

# ------------------------------------------------------------
# 16. PREPARE EVIDENCE
# ------------------------------------------------------------

def prepare_evidence(query, org_id=DEFAULT_ORGANIZATION_ID):
    query = str(query).strip()
    if not query:
        return {"status": "INVALID_QUERY", "evidence": [], "citations": [], "prompt": ""}
    
    if PROMPT_INJECTION_DEFENSE and detect_prompt_injection(query):
        return {"status": "SECURITY_BLOCKED", "evidence": [], "citations": [], "prompt": ""}
    
    results = retrieve_and_rerank(query, org_id)
    secure_results = [r for r in results if r.get("organization_id") == org_id]
    
    if len(secure_results) < MIN_EVIDENCE_COUNT:
        return {"status": "INSUFFICIENT_EVIDENCE", "evidence": [], "citations": [], "prompt": ""}
    
    evidence = []
    citations = []
    for i, result in enumerate(secure_results, 1):
        cid = f"C{i}"
        citations.append({"citation_id": cid, **{k: v for k, v in result.items() if k != "text"}})
        evidence.append({"citation_id": cid, "text": result.get("text", ""), **result})
    
    evidence_blocks = []
    for item in evidence:
        cid = item["citation_id"]
        source = item.get("source_name", "Unknown")
        text = item.get("text", "")
        evidence_blocks.append(f"[{cid}] Source: {source}\nEvidence: {text}")
    
    grounding_context = "\n\n---\n\n".join(evidence_blocks)
    
    prompt = f"""You are SupportIQ, an enterprise AI assistant.

Answer using ONLY the evidence below.

RULES:
1. Only use the evidence provided.
2. Cite sources using [C1], [C2], etc.
3. If evidence doesn't answer, say "I don't have enough information."
4. Be concise and helpful.

QUESTION: {query}

EVIDENCE:
{grounding_context}

ANSWER:"""
    
    return {
        "status": "READY",
        "evidence": evidence,
        "citations": citations,
        "prompt": prompt
    }

# ------------------------------------------------------------
# 17. CALL OPENROUTER
# ------------------------------------------------------------

def call_openrouter(grounded_prompt, query="", org_id=DEFAULT_ORGANIZATION_ID):
    if not OPENROUTER_API_KEY:
        return {"status": "ERROR", "error": "API key missing", "answer": "", "citations": []}
    
    system_message = """You are SupportIQ. Answer using ONLY the evidence provided. Cite sources with [C1], [C2], etc. Be concise."""
    
    try:
        response = requests.post(
            OPENROUTER_API_URL,
            headers={
                "Authorization": f"Bearer {OPENROUTER_API_KEY}",
                "Content-Type": "application/json",
                "HTTP-Referer": "https://supportiq.local",
            },
            json={
                "model": OPENROUTER_MODEL,
                "temperature": LLM_TEMPERATURE,
                "max_tokens": LLM_MAX_TOKENS,
                "messages": [
                    {"role": "system", "content": system_message},
                    {"role": "user", "content": grounded_prompt}
                ]
            },
            timeout=LLM_TIMEOUT_SECONDS
        )
        
        if response.status_code == 200:
            data = response.json()
            content = data["choices"][0]["message"]["content"].strip()
            citations = re.findall(r"\[(C\d+)\]", content)
            return {
                "status": "SUCCESS",
                "answer": content,
                "citations": list(set(citations)),
                "error": None
            }
        else:
            return {"status": "ERROR", "error": f"HTTP {response.status_code}", "answer": "", "citations": []}
    except Exception as e:
        return {"status": "ERROR", "error": str(e), "answer": "", "citations": []}

# ------------------------------------------------------------
# 18. GENERATE ANSWER
# ------------------------------------------------------------

def generate_answer(query, org_id=DEFAULT_ORGANIZATION_ID):
    evidence_result = prepare_evidence(query, org_id)
    if evidence_result.get("status") != "READY":
        return {
            "status": "NO_EVIDENCE",
            "answer": "I don't have enough information to answer that.",
            "citations": [],
            "evidence": evidence_result
        }
    
    llm_result = call_openrouter(evidence_result["prompt"], query, org_id)
    llm_result["evidence"] = evidence_result
    return llm_result

# ------------------------------------------------------------
# 19. TICKET FUNCTIONS
# ------------------------------------------------------------

def load_tickets():
    if TICKETS_JSON.exists():
        with open(TICKETS_JSON, "r", encoding="utf-8") as f:
            data = json.load(f)
            return data if isinstance(data, list) else []
    return []

def save_tickets(tickets):
    with open(TICKETS_JSON, "w", encoding="utf-8") as f:
        json.dump(tickets, f, indent=2, ensure_ascii=False)

def create_ticket(query, org_id=DEFAULT_ORGANIZATION_ID, reason="Human review required",
                  answer="", confidence=0.0, evidence=None):
    if evidence is None:
        evidence = []
    
    ticket = {
        "ticket_id": f"TKT-{uuid.uuid4().hex[:12].upper()}",
        "organization_id": org_id,
        "status": "OPEN",
        "priority": "medium",
        "reason": reason,
        "query": str(query).strip(),
        "ai_answer": str(answer).strip(),
        "confidence": float(confidence),
        "evidence": [{"text": e.get("text", ""), "source": e.get("source_name")} for e in evidence[:5]],
        "created_at": datetime.now(timezone.utc).isoformat(),
        "updated_at": datetime.now(timezone.utc).isoformat()
    }
    
    tickets = load_tickets()
    tickets.append(ticket)
    save_tickets(tickets)
    return ticket

# ------------------------------------------------------------
# 20. MAIN ANSWER FUNCTION
# ------------------------------------------------------------

def supportiq_answer(query, org_id=DEFAULT_ORGANIZATION_ID):
    started = time.time()
    
    if not query or not str(query).strip():
        return {"status": "INVALID_QUERY", "answer": "Please enter a question.", "confidence": 0.0}
    
    query = str(query).strip()
    
    result = generate_answer(query, org_id)
    
    if result.get("status") == "NO_EVIDENCE":
        ticket = create_ticket(query, org_id, "Insufficient evidence", "", 0.0)
        return {
            "status": "ESCALATED",
            "answer": "I don't have enough information to answer. A ticket has been created.",
            "citations": [],
            "confidence": 0.0,
            "ticket_id": ticket.get("ticket_id"),
            "latency": round(time.time() - started, 2)
        }
    
    if result.get("status") == "SUCCESS":
        answer = result.get("answer", "")
        confidence = len(result.get("citations", [])) / 5.0
        confidence = min(1.0, confidence)
        
        if confidence < 0.3:
            ticket = create_ticket(query, org_id, "Low confidence answer", answer, confidence, 
                                   result.get("evidence", {}).get("evidence", []))
            return {
                "status": "ESCALATED",
                "answer": answer + "\n\n(This answer had low confidence and has been escalated.)",
                "citations": result.get("citations", []),
                "confidence": confidence,
                "ticket_id": ticket.get("ticket_id"),
                "latency": round(time.time() - started, 2)
            }
        
        return {
            "status": "ANSWERED",
            "answer": answer,
            "citations": result.get("citations", []),
            "confidence": confidence,
            "ticket_id": None,
            "latency": round(time.time() - started, 2)
        }
    
    return {
        "status": "ERROR",
        "answer": f"Error: {result.get('error', 'Unknown')}",
        "citations": [],
        "confidence": 0.0,
        "latency": round(time.time() - started, 2)
    }

# ------------------------------------------------------------
# 21. BUILD AND STATUS
# ------------------------------------------------------------

def rebuild_knowledge_base(org_id=DEFAULT_ORGANIZATION_ID):
    print("\nRebuilding knowledge base...")
    chunks = build_chunks(org_id)
    embeddings = build_vector_index(org_id)
    print("Knowledge base rebuilt!")
    return {"chunks": len(chunks), "embeddings": len(embeddings)}

def show_status(org_id=DEFAULT_ORGANIZATION_ID):
    docs = load_documents()
    tenant_docs = [d for d in docs if d.get("organization_id") == org_id]
    chunks = load_chunks(org_id)
    embeddings, _ = load_current_index()
    tickets = load_tickets()
    tenant_tickets = [t for t in tickets if t.get("organization_id") == org_id]
    
    print("\n" + "=" * 60)
    print("SUPPORTIQ STATUS")
    print("=" * 60)
    print(f"Organization: {org_id}")
    print(f"Documents: {len(tenant_docs)}")
    print(f"Chunks: {len(chunks)}")
    print(f"Embeddings: {len(embeddings)}")
    print(f"Tickets: {len(tenant_tickets)}")
    print("=" * 60)

# ------------------------------------------------------------
# 22. INITIALIZE
# ------------------------------------------------------------

if not DOCUMENTS_JSON.exists():
    save_documents([])
if not TICKETS_JSON.exists():
    save_tickets([])

print("\n" + "=" * 60)
print("SUPPORTIQ READY")
print("=" * 60)
print("\nHOW TO USE:")
print("  1. Ingest a document:")
print("     ingest_local_file('/path/to/file.pdf')")
print("     ingest_web_url('https://example.com')")
print()
print("  2. Build index:")
print("     rebuild_knowledge_base()")
print()
print("  3. Ask question:")
print("     result = supportiq_answer('what is this about?')")
print("     print(result['answer'])")
print()
print("  4. Check status:")
print("     show_status()")
print("=" * 60)

docs = load_documents()
if docs and not EMBEDDINGS_PATH.exists():
    print("\nBuilding index from existing documents...")
    rebuild_knowledge_base()

print("\n" + "=" * 60)
print("CELL 03 COMPLETE")
print("=" * 60)
```

```python
# ============================================================
# SUPPORTIQ — CELL 04
# WEB APPLICATION
# ============================================================

import os
import time
import traceback
import gradio as gr

print("=" * 60)
print("SUPPORTIQ CELL 04 - WEB APPLICATION")
print("=" * 60)

# ------------------------------------------------------------
# 1. CONFIGURATION
# ------------------------------------------------------------

SUPPORTIQ_ORG = DEFAULT_ORGANIZATION_ID
SUPPORTIQ_MAX_QUERY_LENGTH = MAX_QUERY_LENGTH
SUPPORTIQ_MAX_DOCUMENT_SIZE_MB = MAX_DOCUMENT_SIZE_MB
SUPPORTIQ_MAX_DOCUMENT_SIZE = int(SUPPORTIQ_MAX_DOCUMENT_SIZE_MB * 1024 * 1024)

# ------------------------------------------------------------
# 2. HELPERS
# ------------------------------------------------------------

def safe_text(value, default=""):
    if value is None:
        return default
    return str(value)

def get_system_status():
    try:
        docs = load_documents()
        tenant_docs = [d for d in docs if d.get("organization_id") == SUPPORTIQ_ORG]
        chunks = load_chunks(SUPPORTIQ_ORG)
        tickets = load_tickets()
        tenant_tickets = [t for t in tickets if t.get("organization_id") == SUPPORTIQ_ORG]
        
        return (
            "Organization: " + SUPPORTIQ_ORG + "\n\n"
            "Documents: " + str(len(tenant_docs)) + "\n"
            "Chunks: " + str(len(chunks)) + "\n"
            "Tickets: " + str(len(tenant_tickets)) + "\n\n"
            "SYSTEM COMPONENTS\n"
            "--------------------------------\n"
            "Document ingestion : READY\n"
            "Web ingestion      : READY\n"
            "Chunking           : READY\n"
            "Embeddings         : READY\n"
            "Vector search      : READY\n"
            "Keyword search     : READY\n"
            "Hybrid retrieval   : READY\n"
            "Reranking          : READY\n"
            "Evidence grounding : READY\n"
            "OpenRouter LLM     : READY\n"
            "Verification       : READY\n"
            "Confidence routing : READY\n"
            "Human handoff      : READY"
        )
    except Exception as e:
        return "System status unavailable.\n\nError: " + str(e)

# ------------------------------------------------------------
# 3. REBUILD
# ------------------------------------------------------------

def rebuild_knowledge():
    try:
        result = rebuild_knowledge_base(SUPPORTIQ_ORG)
        return "Knowledge base rebuilt.\n\nChunks: " + str(result.get("chunks", 0)) + "\nEmbeddings: " + str(result.get("embeddings", 0))
    except Exception as e:
        return "Rebuild failed: " + str(e)

# ------------------------------------------------------------
# 4. FILE UPLOAD
# ------------------------------------------------------------

def process_uploaded_file(file_path):
    if not file_path:
        return "Please select a PDF, DOCX, or TXT file.", get_system_status()
    
    file_path = str(file_path)
    print("Uploaded file: " + file_path)
    
    try:
        if not os.path.exists(file_path) or not os.path.isfile(file_path):
            return "Uploaded file could not be found.", get_system_status()
        
        ext = os.path.splitext(file_path)[1].lower()
        if ext not in [".pdf", ".docx", ".txt"]:
            return "Unsupported file type: " + ext + "\n\nAllowed: PDF, DOCX, TXT", get_system_status()
        
        file_size = os.path.getsize(file_path)
        if file_size > SUPPORTIQ_MAX_DOCUMENT_SIZE:
            return "File too large (max " + str(SUPPORTIQ_MAX_DOCUMENT_SIZE_MB) + " MB)", get_system_status()
        
        result = ingest_local_file(file_path, SUPPORTIQ_ORG)
        
        if not isinstance(result, dict):
            return "Unexpected response: " + str(result), get_system_status()
        
        if "already exists" in str(result).lower():
            rebuild_knowledge_base(SUPPORTIQ_ORG)
            return "Document already exists. Knowledge base rebuilt.", get_system_status()
        
        rebuild_knowledge_base(SUPPORTIQ_ORG)
        
        return (
            "DOCUMENT READY\n\n"
            "File uploaded\n"
            "Text extracted\n"
            "Document chunked\n"
            "Embeddings generated\n"
            "Ready for questions",
            get_system_status()
        )
        
    except Exception as e:
        print("Error: " + str(e))
        traceback.print_exc()
        return "Upload failed: " + str(e), get_system_status()

# ------------------------------------------------------------
# 5. WEB INGESTION
# ------------------------------------------------------------

def process_web_url(url):
    if not url or not str(url).strip():
        return "Please enter a web URL.", get_system_status()
    
    url = str(url).strip()
    if not (url.startswith("http://") or url.startswith("https://")):
        return "Invalid URL. Use http:// or https://", get_system_status()
    
    print("Web URL: " + url)
    
    try:
        result = ingest_web_url(url, SUPPORTIQ_ORG)
        
        if not isinstance(result, dict):
            return "Unexpected response: " + str(result), get_system_status()
        
        if "already exists" in str(result).lower():
            rebuild_knowledge_base(SUPPORTIQ_ORG)
            return "Web page already exists. Knowledge base rebuilt.", get_system_status()
        
        rebuild_knowledge_base(SUPPORTIQ_ORG)
        
        return (
            "WEB PAGE READY\n\n"
            "Page downloaded\n"
            "Content extracted\n"
            "Chunks created\n"
            "Embeddings generated\n"
            "Ready for questions",
            get_system_status()
        )
        
    except Exception as e:
        print("Error: " + str(e))
        traceback.print_exc()
        return "Web ingestion failed: " + str(e), get_system_status()

# ------------------------------------------------------------
# 6. CHAT FUNCTION
# ------------------------------------------------------------

def chat_with_supportiq(query):
    if not query or not str(query).strip():
        return "Please enter a customer question.", "", ""
    
    query = str(query).strip()
    if len(query) > SUPPORTIQ_MAX_QUERY_LENGTH:
        return "Question too long (max " + str(SUPPORTIQ_MAX_QUERY_LENGTH) + " chars)", "", ""
    
    print("Question: " + query)
    start_time = time.time()
    
    try:
        result = supportiq_answer(query, SUPPORTIQ_ORG)
        latency = time.time() - start_time
        
        answer = safe_text(result.get("answer", "No answer generated."))
        status = safe_text(result.get("status", "UNKNOWN"))
        confidence = float(result.get("confidence", 0.0))
        citations = result.get("citations", [])
        ticket_id = safe_text(result.get("ticket_id", ""))
        
        # Format citations
        citation_text = "No citations returned."
        if citations:
            citation_lines = []
            for c in citations:
                citation_lines.append("  " + str(c))
            citation_text = "\n".join(citation_lines)
        
        # Routing info
        if status == "ANSWERED":
            routing_text = (
                "ANSWERED\n\n"
                "Confidence: " + str(round(confidence * 100, 1)) + "%\n"
                "State: VERIFIED\n"
                "Action: RETURN_ANSWER\n"
                "Latency: " + str(round(latency, 2)) + "s"
            )
        elif status == "ESCALATED":
            routing_text = (
                "HUMAN HANDOFF REQUIRED\n\n"
                "Confidence: " + str(round(confidence * 100, 1)) + "%\n"
                "State: ESCALATE\n"
            )
            if ticket_id:
                routing_text += "Ticket ID: " + ticket_id + "\n"
            routing_text += "\nThe AI did not have sufficient verified evidence to safely answer."
        else:
            routing_text = "Status: " + status + "\nConfidence: " + str(round(confidence * 100, 1)) + "%"
        
        return answer, citation_text, routing_text
        
    except Exception as e:
        print("Error: " + str(e))
        traceback.print_exc()
        return "SupportIQ could not process this request.", "", "ERROR: " + str(e)

# ------------------------------------------------------------
# 7. TICKETS DISPLAY
# ------------------------------------------------------------

def show_tickets():
    try:
        tickets = load_tickets()
        tenant_tickets = [t for t in tickets if t.get("organization_id") == SUPPORTIQ_ORG]
        
        if not tenant_tickets:
            return "No support tickets currently exist."
        
        rows = []
        for ticket in tenant_tickets:
            rows.append(
                "TICKET\n"
                "----------------------------------------\n"
                "ID: " + safe_text(ticket.get("ticket_id")) + "\n"
                "Status: " + safe_text(ticket.get("status")) + "\n"
                "Priority: " + safe_text(ticket.get("priority")) + "\n"
                "Confidence: " + safe_text(ticket.get("confidence")) + "\n"
                "Question: " + safe_text(ticket.get("query")) + "\n"
                "Reason: " + safe_text(ticket.get("reason")) + "\n"
            )
        return "\n".join(rows)
    except Exception as e:
        return "Could not load tickets: " + str(e)

# ------------------------------------------------------------
# 8. GRADIO INTERFACE
# ------------------------------------------------------------

print("\nBuilding Gradio interface...")

with gr.Blocks(title="SupportIQ", theme=gr.themes.Soft()) as demo:
    
    gr.Markdown("""
    # SupportIQ
    ### Enterprise AI Customer-Support Copilot
    """)
    
    with gr.Row():
        with gr.Column(scale=2):
            gr.Markdown("### Ask a Question")
            customer_question = gr.Textbox(
                label="Customer Question",
                placeholder="Example: What is the return policy?",
                lines=4
            )
            
            with gr.Row():
                ask_button = gr.Button("Ask SupportIQ", variant="primary", size="lg")
                clear_button = gr.Button("Clear", size="lg")
            
            answer_box = gr.Textbox(label="Answer", lines=8, interactive=False)
            citation_box = gr.Textbox(label="Sources", lines=4, interactive=False)
            routing_box = gr.Textbox(label="Confidence / Routing", lines=6, interactive=False)
        
        with gr.Column(scale=1):
            gr.Markdown("### System Status")
            knowledge_status = gr.Textbox(
                label="Knowledge Base",
                value=get_system_status(),
                lines=18,
                interactive=False
            )
            refresh_status = gr.Button("Refresh Status", size="sm")
            refresh_status.click(fn=get_system_status, outputs=knowledge_status)
    
    gr.Markdown("---")
    gr.Markdown("### Upload Knowledge")
    
    with gr.Row():
        with gr.Column():
            gr.Markdown("**Documents**")
            file_upload = gr.File(
                label="Upload PDF / DOCX / TXT",
                file_types=[".pdf", ".docx", ".txt"],
                file_count="single",
                type="filepath"
            )
            upload_button = gr.Button("Process Document", variant="primary")
            upload_status = gr.Textbox(label="Upload Status", lines=4, interactive=False)
        
        with gr.Column():
            gr.Markdown("**Web Pages**")
            web_url = gr.Textbox(label="Web URL", placeholder="https://example.com/help")
            web_button = gr.Button("Add Web Page", variant="primary")
            web_status = gr.Textbox(label="Web Status", lines=4, interactive=False)
    
    gr.Markdown("---")
    gr.Markdown("### Support Tickets")
    with gr.Row():
        tickets_box = gr.Textbox(label="Tickets", lines=10, interactive=False)
        refresh_tickets = gr.Button("Refresh Tickets", size="sm")
        refresh_tickets.click(fn=show_tickets, outputs=tickets_box)
    
    # Connect buttons
    ask_button.click(
        fn=chat_with_supportiq,
        inputs=customer_question,
        outputs=[answer_box, citation_box, routing_box]
    )
    
    customer_question.submit(
        fn=chat_with_supportiq,
        inputs=customer_question,
        outputs=[answer_box, citation_box, routing_box]
    )
    
    upload_button.click(
        fn=process_uploaded_file,
        inputs=file_upload,
        outputs=[upload_status, knowledge_status]
    )
    
    web_button.click(
        fn=process_web_url,
        inputs=web_url,
        outputs=[web_status, knowledge_status]
    )
    
    clear_button.click(
        fn=lambda: ("", "", "", ""),
        outputs=[customer_question, answer_box, citation_box, routing_box]
    )
    
    # Load tickets on startup
    tickets_box.value = show_tickets()

# ------------------------------------------------------------
# 9. LAUNCH
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("SUPPORTIQ APPLICATION READY")
print("=" * 60)
print("Organization: " + SUPPORTIQ_ORG)
print("")
print("Features:")
print("  Ask questions with cited answers")
print("  Upload PDF, DOCX, TXT files")
print("  Add web pages")
print("  View support tickets")
print("=" * 60)

demo.queue()
demo.launch(share=True, debug=False)

print("\n" + "=" * 60)
print("CELL 04 COMPLETE")
print("=" * 60)
```

```python
# ============================================================
# SUPPORTIQ — CELL 03 (COMPLETE FIXED)
# ============================================================

from pathlib import Path
from urllib.parse import urlparse
import hashlib
import json
import re
import shutil
import uuid
import html
import time
import math
from datetime import datetime, timezone
from rank_bm25 import BM25Okapi
from pypdf import PdfReader
from docx import Document

print("=" * 60)
print("SUPPORTIQ CELL 03 - ALL FUNCTIONS (FIXED)")
print("=" * 60)

# ------------------------------------------------------------
# FILE PATHS
# ------------------------------------------------------------

DOCUMENTS_JSON = DATA_DIR / "documents.json"
CHUNKS_JSON = DATA_DIR / "chunks.json"
EMBEDDINGS_PATH = INDEX_DIR / "embeddings.npy"
METADATA_PATH = INDEX_DIR / "chunk_metadata.json"
TICKETS_JSON = TICKET_DIR / "tickets.json"

# ------------------------------------------------------------
# TEXT CLEANING
# ------------------------------------------------------------

def clean_text(text):
    if not text:
        return ""
    text = html.unescape(str(text))
    text = text.replace("\r\n", "\n").replace("\r", "\n")
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n{3,}", "\n\n", text)
    return text.strip()

def estimate_tokens(text):
    return len(re.findall(r"\S+", text))

def split_sentences(text):
    text = clean_text(text)
    if not text:
        return []
    abbr = {"e.g.": "eg__", "i.e.": "ie__", "Mr.": "Mr__", "Mrs.": "Mrs__", "Dr.": "Dr__"}
    for old, new in abbr.items():
        text = text.replace(old, new)
    sentences = re.split(r"(?<=[.!?])\s+|\n{2,}", text)
    result = []
    for s in sentences:
        s = s.strip()
        if s:
            for old, new in abbr.items():
                s = s.replace(new, old)
            result.append(s)
    return result

# ------------------------------------------------------------
# CHUNKING
# ------------------------------------------------------------

def chunk_text(text):
    sentences = split_sentences(text)
    if not sentences:
        return []
    chunks = []
    current = []
    current_tokens = 0
    for sentence in sentences:
        sent_tokens = estimate_tokens(sentence)
        if sent_tokens > CHUNK_SIZE:
            if current:
                chunks.append(" ".join(current))
                current = []
                current_tokens = 0
            tokens = re.findall(r"\S+", sentence)
            start = 0
            while start < len(tokens):
                end = min(start + CHUNK_SIZE, len(tokens))
                piece = " ".join(tokens[start:end])
                if piece.strip():
                    chunks.append(piece)
                start = end - CHUNK_OVERLAP if end - CHUNK_OVERLAP > start else end
            continue
        if current_tokens + sent_tokens <= CHUNK_SIZE:
            current.append(sentence)
            current_tokens += sent_tokens
        else:
            if current:
                chunks.append(" ".join(current))
            overlap_sentences = []
            overlap_tokens = 0
            for prev in reversed(current):
                prev_tokens = estimate_tokens(prev)
                if overlap_tokens + prev_tokens > CHUNK_OVERLAP:
                    break
                overlap_sentences.insert(0, prev)
                overlap_tokens += prev_tokens
            current = overlap_sentences + [sentence]
            current_tokens = overlap_tokens + sent_tokens
    if current:
        chunks.append(" ".join(current))
    return [c for c in chunks if len(c) >= MIN_CHUNK_CHARACTERS]

# ------------------------------------------------------------
# FILE EXTRACTION
# ------------------------------------------------------------

def extract_pdf(file_path):
    pages = []
    reader = PdfReader(str(file_path))
    for i, page in enumerate(reader.pages, 1):
        try:
            text = clean_text(page.extract_text() or "")
            if text:
                pages.append({"page": i, "text": text})
        except:
            continue
    return "\n\n".join(p["text"] for p in pages), pages

def extract_docx(file_path):
    doc = Document(str(file_path))
    sections = []
    for para in doc.paragraphs:
        text = clean_text(para.text)
        if text:
            sections.append(text)
    for table in doc.tables:
        for row in table.rows:
            cells = [clean_text(c.text) for c in row.cells if clean_text(c.text)]
            if cells:
                sections.append(" | ".join(cells))
    return "\n\n".join(sections), []

def extract_txt(file_path):
    for enc in ["utf-8", "utf-8-sig", "cp1252"]:
        try:
            with open(file_path, "r", encoding=enc) as f:
                return clean_text(f.read()), []
        except:
            continue
    with open(file_path, "r", encoding="utf-8", errors="replace") as f:
        return clean_text(f.read()), []

def extract_file(file_path):
    ext = file_path.suffix.lower()
    if ext == ".pdf":
        return extract_pdf(file_path)
    elif ext == ".docx":
        return extract_docx(file_path)
    elif ext == ".txt":
        return extract_txt(file_path)
    raise ValueError(f"Unsupported: {ext}")

# ------------------------------------------------------------
# FILE HASH
# ------------------------------------------------------------

def calculate_sha256(file_path):
    sha = hashlib.sha256()
    with open(file_path, "rb") as f:
        for block in iter(lambda: f.read(65536), b""):
            sha.update(block)
    return sha.hexdigest()

# ------------------------------------------------------------
# DOCUMENT OPERATIONS
# ------------------------------------------------------------

def load_documents():
    if DOCUMENTS_JSON.exists():
        with open(DOCUMENTS_JSON, "r", encoding="utf-8") as f:
            data = json.load(f)
            return data if isinstance(data, list) else []
    return []

def save_documents(docs):
    with open(DOCUMENTS_JSON, "w", encoding="utf-8") as f:
        json.dump(docs, f, indent=2, ensure_ascii=False)

# ------------------------------------------------------------
# INGEST LOCAL FILE
# ------------------------------------------------------------

def ingest_local_file(file_path, org_id=DEFAULT_ORGANIZATION_ID):
    print(f"\nIngesting: {file_path}")
    source = Path(str(file_path)).resolve()
    if not source.exists():
        raise FileNotFoundError(f"File not found: {source}")
    
    ext = source.suffix.lower()
    if ext not in [".pdf", ".docx", ".txt"]:
        raise ValueError(f"Unsupported: {ext}")
    
    size_mb = source.stat().st_size / (1024 * 1024)
    if size_mb > MAX_DOCUMENT_SIZE_MB:
        raise ValueError(f"File exceeds {MAX_DOCUMENT_SIZE_MB}MB")
    
    file_hash = calculate_sha256(source)
    
    docs = load_documents()
    for doc in docs:
        if doc.get("file_hash") == file_hash:
            print("Document already exists")
            return doc
    
    safe_name = re.sub(r"[^a-zA-Z0-9._-]+", "_", source.name)
    dest = DOCUMENT_DIR / f"{file_hash[:12]}_{safe_name}"
    shutil.copy2(source, dest)
    
    text, pages = extract_file(dest)
    if not text.strip():
        raise ValueError("No readable text extracted")
    
    doc_id = f"DOC-{uuid.uuid4().hex[:12].upper()}"
    record = {
        "document_id": doc_id,
        "organization_id": org_id,
        "source_type": "file",
        "source_name": source.name,
        "source_path": str(dest),
        "file_hash": file_hash,
        "text": text,
        "pages": pages,
        "character_count": len(text),
        "created_at": datetime.now(timezone.utc).isoformat(),
        "status": "READY"
    }
    
    docs.append(record)
    save_documents(docs)
    print(f"Document ingested: {doc_id}")
    return record

# ------------------------------------------------------------
# INGEST WEB
# ------------------------------------------------------------

def ingest_web_url(url, org_id=DEFAULT_ORGANIZATION_ID):
    print(f"\nIngesting web: {url}")
    
    parsed = urlparse(url)
    if parsed.scheme.lower() not in ["http", "https"]:
        raise ValueError("Only HTTP/HTTPS supported")
    
    response = requests.get(url, timeout=20, headers={"User-Agent": "SupportIQ/2.0"})
    response.raise_for_status()
    
    soup = BeautifulSoup(response.text, "lxml")
    for tag in soup(["script", "style", "noscript", "svg", "nav", "footer", "header"]):
        tag.decompose()
    
    title = clean_text(soup.title.get_text(" ", strip=True)) if soup.title else url
    main = soup.find("main") or soup.find("article") or soup.body or soup
    text = clean_text(main.get_text("\n", strip=True))
    
    if not text:
        raise ValueError("No readable text found")
    
    content_hash = hashlib.sha256((url + "\n" + text).encode()).hexdigest()
    
    docs = load_documents()
    for doc in docs:
        if doc.get("file_hash") == content_hash:
            print("Web page already exists")
            return doc
    
    doc_id = f"WEB-{uuid.uuid4().hex[:12].upper()}"
    record = {
        "document_id": doc_id,
        "organization_id": org_id,
        "source_type": "web",
        "source_name": title,
        "source_url": response.url,
        "file_hash": content_hash,
        "text": text,
        "pages": [],
        "character_count": len(text),
        "created_at": datetime.now(timezone.utc).isoformat(),
        "status": "READY"
    }
    
    docs.append(record)
    save_documents(docs)
    print(f"Web page ingested: {doc_id}")
    return record

# ------------------------------------------------------------
# BUILD CHUNKS
# ------------------------------------------------------------

def build_chunks(org_id=DEFAULT_ORGANIZATION_ID):
    print(f"\nBuilding chunks for: {org_id}")
    
    docs = load_documents()
    tenant_docs = [d for d in docs if d.get("organization_id") == org_id]
    
    if not tenant_docs:
        print("No documents found")
        save_chunks([])
        return []
    
    all_chunks = []
    for doc in tenant_docs:
        doc_id = doc.get("document_id")
        print(f"Processing: {doc.get('source_name')}")
        
        text_chunks = chunk_text(doc.get("text", ""))
        for i, chunk in enumerate(text_chunks, 1):
            chunk_id = f"{doc_id}-C{i:04d}"
            all_chunks.append({
                "chunk_id": chunk_id,
                "document_id": doc_id,
                "organization_id": org_id,
                "chunk_number": i,
                "text": chunk,
                "source_type": doc.get("source_type"),
                "source_name": doc.get("source_name"),
                "source_url": doc.get("source_url"),
                "page": None,
                "character_count": len(chunk),
                "token_estimate": estimate_tokens(chunk)
            })
        
        print(f"  Created {len(text_chunks)} chunks")
    
    save_chunks(all_chunks)
    print(f"Total chunks: {len(all_chunks)}")
    return all_chunks

def save_chunks(chunks):
    with open(CHUNKS_JSON, "w", encoding="utf-8") as f:
        json.dump(chunks, f, indent=2, ensure_ascii=False)

def load_chunks(org_id=None):
    if not CHUNKS_JSON.exists():
        return []
    with open(CHUNKS_JSON, "r", encoding="utf-8") as f:
        data = json.load(f)
    if org_id:
        return [c for c in data if c.get("organization_id") == org_id]
    return data

# ------------------------------------------------------------
# BUILD VECTOR INDEX
# ------------------------------------------------------------

def build_vector_index(org_id=DEFAULT_ORGANIZATION_ID):
    print(f"\nBuilding vector index for: {org_id}")
    
    chunks = load_chunks(org_id)
    if not chunks:
        print("No chunks available")
        np.save(EMBEDDINGS_PATH, np.empty((0, EMBEDDING_DIMENSION), dtype=np.float32))
        with open(METADATA_PATH, "w", encoding="utf-8") as f:
            json.dump([], f, indent=2)
        return np.empty((0, EMBEDDING_DIMENSION), dtype=np.float32)
    
    texts = [c.get("text", "") for c in chunks]
    valid = [i for i, t in enumerate(texts) if t and t.strip()]
    chunks = [chunks[i] for i in valid]
    texts = [texts[i] for i in valid]
    
    print(f"Generating embeddings for {len(texts)} chunks...")
    embeddings = embedding_model.encode(
        texts,
        batch_size=EMBEDDING_BATCH_SIZE,
        show_progress_bar=True,
        convert_to_numpy=True,
        normalize_embeddings=True
    )
    embeddings = np.asarray(embeddings, dtype=np.float32)
    
    np.save(EMBEDDINGS_PATH, embeddings)
    
    metadata = []
    for chunk in chunks:
        metadata.append({
            "chunk_id": chunk.get("chunk_id"),
            "document_id": chunk.get("document_id"),
            "organization_id": chunk.get("organization_id"),
            "text": chunk.get("text"),
            "source_type": chunk.get("source_type"),
            "source_name": chunk.get("source_name"),
            "source_url": chunk.get("source_url"),
            "page": chunk.get("page"),
            "chunk_number": chunk.get("chunk_number"),
        })
    
    with open(METADATA_PATH, "w", encoding="utf-8") as f:
        json.dump(metadata, f, indent=2, ensure_ascii=False)
    
    print(f"Vector index built: {embeddings.shape[0]} vectors")
    return embeddings

# ------------------------------------------------------------
# LOAD INDEX
# ------------------------------------------------------------

def load_vector_index():
    if not EMBEDDINGS_PATH.exists():
        return np.empty((0, EMBEDDING_DIMENSION), dtype=np.float32)
    return np.load(EMBEDDINGS_PATH)

def load_metadata():
    if not METADATA_PATH.exists():
        return []
    with open(METADATA_PATH, "r", encoding="utf-8") as f:
        return json.load(f)

def load_current_index():
    embeddings = load_vector_index()
    metadata = load_metadata()
    min_len = min(len(embeddings), len(metadata))
    return embeddings[:min_len], metadata[:min_len]

# ------------------------------------------------------------
# BM25 TOKENS
# ------------------------------------------------------------

def bm25_tokens(text):
    return re.findall(r"\b[\w+#.-]+\b", str(text).lower())

# ------------------------------------------------------------
# HYBRID SEARCH
# ------------------------------------------------------------

def reciprocal_rank_fusion(vector_results, keyword_results, k=60):
    scores = {}
    result_map = {}
    for rank, item in enumerate(vector_results, 1):
        chunk_id = item["chunk_id"]
        scores[chunk_id] = scores.get(chunk_id, 0.0) + 1.0 / (k + rank)
        result_map[chunk_id] = item
    for rank, item in enumerate(keyword_results, 1):
        chunk_id = item["chunk_id"]
        scores[chunk_id] = scores.get(chunk_id, 0.0) + 1.0 / (k + rank)
        result_map[chunk_id] = item
    fused = []
    for chunk_id, score in scores.items():
        item = dict(result_map[chunk_id])
        item["rrf_score"] = float(score)
        fused.append(item)
    fused.sort(key=lambda x: x["rrf_score"], reverse=True)
    return fused

def hybrid_search(query, org_id=DEFAULT_ORGANIZATION_ID):
    query = str(query).strip()
    if not query:
        return []
    
    embeddings, metadata = load_current_index()
    allowed = [i for i, item in enumerate(metadata) 
               if item.get("organization_id") == org_id]
    if not allowed:
        return []
    
    # Vector search
    q_emb = embedding_model.encode([query], convert_to_numpy=True, normalize_embeddings=True)[0]
    tenant_embeddings = embeddings[allowed]
    scores = tenant_embeddings @ q_emb
    ranking = np.argsort(scores)[::-1][:VECTOR_TOP_K]
    vector_results = []
    for rank_pos in ranking:
        original_idx = allowed[int(rank_pos)]
        item = dict(metadata[original_idx])
        item["vector_score"] = float(scores[rank_pos])
        vector_results.append(item)
    
    # BM25 search
    tenant_metadata = [metadata[i] for i in allowed]
    corpus = [bm25_tokens(item.get("text", "")) for item in tenant_metadata]
    query_tokens = bm25_tokens(query)
    if query_tokens:
        bm25 = BM25Okapi(corpus)
        bm25_scores = bm25.get_scores(query_tokens)
        ranking = np.argsort(bm25_scores)[::-1][:BM25_TOP_K]
        keyword_results = []
        for idx in ranking:
            idx = int(idx)
            item = dict(tenant_metadata[idx])
            item["keyword_score"] = float(bm25_scores[idx])
            keyword_results.append(item)
    else:
        keyword_results = []
    
    fused = reciprocal_rank_fusion(vector_results, keyword_results)
    return fused[:HYBRID_TOP_K]

# ------------------------------------------------------------
# TICKET FUNCTIONS
# ------------------------------------------------------------

def load_tickets():
    if TICKETS_JSON.exists():
        with open(TICKETS_JSON, "r", encoding="utf-8") as f:
            data = json.load(f)
            return data if isinstance(data, list) else []
    return []

def save_tickets(tickets):
    with open(TICKETS_JSON, "w", encoding="utf-8") as f:
        json.dump(tickets, f, indent=2, ensure_ascii=False)

def create_ticket(query, org_id=DEFAULT_ORGANIZATION_ID, reason="Human review required",
                  answer="", confidence=0.0, evidence=None):
    if evidence is None:
        evidence = []
    
    ticket = {
        "ticket_id": f"TKT-{uuid.uuid4().hex[:12].upper()}",
        "organization_id": org_id,
        "status": "OPEN",
        "priority": "medium",
        "reason": reason,
        "query": str(query).strip(),
        "ai_answer": str(answer).strip(),
        "confidence": float(confidence),
        "evidence": [{"text": e.get("text", ""), "source": e.get("source_name")} for e in evidence[:5]],
        "created_at": datetime.now(timezone.utc).isoformat(),
        "updated_at": datetime.now(timezone.utc).isoformat()
    }
    
    tickets = load_tickets()
    tickets.append(ticket)
    save_tickets(tickets)
    return ticket

# ------------------------------------------------------------
# MAIN ANSWER FUNCTION (FIXED)
# ------------------------------------------------------------

def supportiq_answer(query, org_id=DEFAULT_ORGANIZATION_ID):
    """Main function - uses hybrid search directly"""
    started = time.time()
    
    if not query or not str(query).strip():
        return {"status": "INVALID_QUERY", "answer": "Please enter a question.", "confidence": 0.0}
    
    query = str(query).strip()
    
    # Get search results directly
    results = hybrid_search(query, org_id)
    
    if not results:
        ticket = create_ticket(query, org_id, "No results found", "", 0.0)
        return {
            "status": "ESCALATED",
            "answer": "I couldn't find any relevant information in the knowledge base.",
            "citations": [],
            "confidence": 0.0,
            "ticket_id": ticket.get("ticket_id"),
            "latency": round(time.time() - started, 2)
        }
    
    # Filter by org
    results = [r for r in results if r.get("organization_id") == org_id]
    
    if not results:
        ticket = create_ticket(query, org_id, "No results found", "", 0.0)
        return {
            "status": "ESCALATED",
            "answer": "I couldn't find any relevant information in the knowledge base.",
            "citations": [],
            "confidence": 0.0,
            "ticket_id": ticket.get("ticket_id"),
            "latency": round(time.time() - started, 2)
        }
    
    # Build context from top results
    context_parts = []
    citations = []
    for i, r in enumerate(results[:5], 1):
        cid = f"C{i}"
        citations.append(cid)
        context_parts.append(f"[{cid}] {r.get('text', '')}")
    
    context = "\n\n---\n\n".join(context_parts)
    
    # Build prompt
    prompt = f"""You are SupportIQ. Answer the question using ONLY the information below.

QUESTION: {query}

INFORMATION:
{context}

RULES:
1. Only use the information above.
2. Cite sources using [C1], [C2], etc.
3. Be concise and helpful.
4. If the information doesn't answer the question, say so.

ANSWER:"""
    
    # Call LLM
    try:
        response = requests.post(
            OPENROUTER_API_URL,
            headers={
                "Authorization": f"Bearer {OPENROUTER_API_KEY}",
                "Content-Type": "application/json",
                "HTTP-Referer": "https://supportiq.local",
            },
            json={
                "model": OPENROUTER_MODEL,
                "temperature": LLM_TEMPERATURE,
                "max_tokens": LLM_MAX_TOKENS,
                "messages": [
                    {"role": "system", "content": "You are SupportIQ. Answer using ONLY the provided information. Cite sources."},
                    {"role": "user", "content": prompt}
                ]
            },
            timeout=LLM_TIMEOUT_SECONDS
        )
        
        if response.status_code == 200:
            data = response.json()
            answer = data["choices"][0]["message"]["content"].strip()
            
            # Extract citations from answer
            found_citations = re.findall(r"\[(C\d+)\]", answer)
            used_citations = list(set(found_citations))
            
            # Calculate confidence based on citations
            confidence = min(1.0, len(used_citations) / 3.0 + 0.2)
            
            return {
                "status": "ANSWERED",
                "answer": answer,
                "citations": used_citations,
                "confidence": round(confidence, 2),
                "ticket_id": None,
                "latency": round(time.time() - started, 2)
            }
        else:
            return {
                "status": "ERROR",
                "answer": f"API Error: {response.status_code}",
                "citations": [],
                "confidence": 0.0,
                "latency": round(time.time() - started, 2)
            }
    except Exception as e:
        return {
            "status": "ERROR",
            "answer": f"Error: {str(e)}",
            "citations": [],
            "confidence": 0.0,
            "latency": round(time.time() - started, 2)
        }

# ------------------------------------------------------------
# REBUILD AND STATUS
# ------------------------------------------------------------

def rebuild_knowledge_base(org_id=DEFAULT_ORGANIZATION_ID):
    print("\nRebuilding knowledge base...")
    chunks = build_chunks(org_id)
    embeddings = build_vector_index(org_id)
    print("Knowledge base rebuilt!")
    return {"chunks": len(chunks), "embeddings": len(embeddings)}

def show_status(org_id=DEFAULT_ORGANIZATION_ID):
    docs = load_documents()
    tenant_docs = [d for d in docs if d.get("organization_id") == org_id]
    chunks = load_chunks(org_id)
    embeddings, _ = load_current_index()
    tickets = load_tickets()
    tenant_tickets = [t for t in tickets if t.get("organization_id") == org_id]
    
    print("\n" + "=" * 60)
    print("SUPPORTIQ STATUS")
    print("=" * 60)
    print(f"Organization: {org_id}")
    print(f"Documents: {len(tenant_docs)}")
    print(f"Chunks: {len(chunks)}")
    print(f"Embeddings: {len(embeddings)}")
    print(f"Tickets: {len(tenant_tickets)}")
    print("=" * 60)

# ------------------------------------------------------------
# INITIALIZE
# ------------------------------------------------------------

if not DOCUMENTS_JSON.exists():
    save_documents([])
if not TICKETS_JSON.exists():
    save_tickets([])

print("\n" + "=" * 60)
print("SUPPORTIQ READY")
print("=" * 60)
print("\nHOW TO USE:")
print("  1. Ingest a document:")
print("     ingest_local_file('/path/to/file.pdf')")
print("     ingest_web_url('https://example.com')")
print()
print("  2. Build index:")
print("     rebuild_knowledge_base()")
print()
print("  3. Ask question:")
print("     result = supportiq_answer('what is this about?')")
print("     print(result['answer'])")
print()
print("  4. Check status:")
print("     show_status()")
print("=" * 60)

# Auto-build if documents exist and index missing
docs = load_documents()
if docs and not EMBEDDINGS_PATH.exists():
    print("\nBuilding index from existing documents...")
    rebuild_knowledge_base()

print("\n" + "=" * 60)
print("CELL 03 COMPLETE")
print("=" * 60)
```

```python
# ============================================================
# SUPPORTIQ — CELL 04
# COMPLETE WEB SYSTEM (GRADIO)
# ============================================================

import gradio as gr
import os
import time
import tempfile
from pathlib import Path

print("=" * 60)
print("SUPPORTIQ - COMPLETE WEB SYSTEM")
print("=" * 60)

# ------------------------------------------------------------
# CORE FUNCTIONS
# ------------------------------------------------------------

def add_documents(file_paths):
    """Add multiple documents"""
    if not file_paths:
        return "No files selected.", get_status()
    
    results = []
    for file_path in file_paths:
        try:
            result = ingest_local_file(file_path, DEFAULT_ORGANIZATION_ID)
            if isinstance(result, dict):
                results.append(f"✓ {result.get('source_name')} - {result.get('document_id')}")
            else:
                results.append(f"✗ {os.path.basename(file_path)} - {result}")
        except Exception as e:
            results.append(f"✗ {os.path.basename(file_path)} - {str(e)}")
    
    # Rebuild index after adding all
    rebuild_knowledge_base(DEFAULT_ORGANIZATION_ID)
    
    return "\n".join(results), get_status()

def add_web_urls(urls):
    """Add multiple web URLs"""
    if not urls:
        return "No URLs entered.", get_status()
    
    results = []
    for url in urls.strip().split():
        url = url.strip()
        if not url:
            continue
        try:
            result = ingest_web_url(url, DEFAULT_ORGANIZATION_ID)
            if isinstance(result, dict):
                results.append(f"✓ {url} - {result.get('document_id')}")
            else:
                results.append(f"✗ {url} - {result}")
        except Exception as e:
            results.append(f"✗ {url} - {str(e)}")
    
    # Rebuild index
    rebuild_knowledge_base(DEFAULT_ORGANIZATION_ID)
    
    return "\n".join(results), get_status()

def chat_with_knowledge(query):
    """Answer questions using knowledge base"""
    if not query or not query.strip():
        return "Please enter a question.", "", ""
    
    result = supportiq_answer(query.strip(), DEFAULT_ORGANIZATION_ID)
    
    answer = result.get("answer", "No answer generated.")
    citations = result.get("citations", [])
    status = result.get("status", "UNKNOWN")
    confidence = result.get("confidence", 0.0)
    latency = result.get("latency", 0)
    ticket_id = result.get("ticket_id")
    
    citation_text = "No citations."
    if citations:
        citation_text = "\n".join([f"• {c}" for c in citations])
    
    status_text = f"""Status: {status}
Confidence: {confidence:.1%}
Latency: {latency:.2f}s"""
    
    if ticket_id:
        status_text += f"\nTicket: {ticket_id}"
    
    return answer, citation_text, status_text

def clear_chat():
    return "", "", ""

def get_status():
    """Get system status"""
    docs = load_documents()
    tenant_docs = [d for d in docs if d.get("organization_id") == DEFAULT_ORGANIZATION_ID]
    chunks = load_chunks(DEFAULT_ORGANIZATION_ID)
    embeddings, _ = load_current_index()
    tickets = load_tickets()
    tenant_tickets = [t for t in tickets if t.get("organization_id") == DEFAULT_ORGANIZATION_ID]
    
    return f"""=== SYSTEM STATUS ===

Organization: {DEFAULT_ORGANIZATION_ID}

📄 DOCUMENTS: {len(tenant_docs)}
   {', '.join([d.get('source_name', '')[:30] for d in tenant_docs[:3]])}{'...' if len(tenant_docs) > 3 else ''}

📦 CHUNKS: {len(chunks)}

🔢 EMBEDDINGS: {len(embeddings)}

🎫 TICKETS: {len(tenant_tickets)}

🟢 STATUS: READY
"""

# ------------------------------------------------------------
# BUILD GRADIO INTERFACE
# ------------------------------------------------------------

with gr.Blocks(title="SupportIQ", theme=gr.themes.Soft()) as demo:
    
    gr.Markdown("""
    # 🤖 SupportIQ
    ### Enterprise AI Customer-Support Copilot
    
    Upload documents, add web pages, and ask questions with cited answers.
    """)
    
    with gr.Tabs():
        
        # TAB 1: Chat
        with gr.TabItem("💬 Ask Questions"):
            with gr.Row():
                with gr.Column(scale=3):
                    gr.Markdown("### Ask a Question")
                    question = gr.Textbox(
                        label="Your Question",
                        placeholder="Ask anything about your uploaded documents...",
                        lines=3
                    )
                    with gr.Row():
                        ask_btn = gr.Button("Ask", variant="primary", size="lg")
                        clear_btn = gr.Button("Clear", size="lg")
                    
                    answer = gr.Textbox(label="Answer", lines=8, interactive=False)
                    citations = gr.Textbox(label="Citations", lines=4, interactive=False)
                    status = gr.Textbox(label="Status", lines=4, interactive=False)
                
                with gr.Column(scale=1):
                    gr.Markdown("### System Status")
                    status_box = gr.Textbox(value=get_status(), lines=20, interactive=False)
                    refresh_btn = gr.Button("Refresh Status")
                    refresh_btn.click(fn=get_status, outputs=status_box)
            
            ask_btn.click(
                fn=chat_with_knowledge,
                inputs=question,
                outputs=[answer, citations, status]
            )
            question.submit(
                fn=chat_with_knowledge,
                inputs=question,
                outputs=[answer, citations, status]
            )
            clear_btn.click(
                fn=clear_chat,
                outputs=[question, answer, citations]
            )
        
        # TAB 2: Upload Documents
        with gr.TabItem("📄 Upload Documents"):
            gr.Markdown("### Upload Multiple Documents")
            gr.Markdown("Supported: PDF, DOCX, TXT")
            
            with gr.Row():
                with gr.Column():
                    file_upload = gr.File(
                        label="Select Documents",
                        file_types=[".pdf", ".docx", ".txt"],
                        file_count="multiple",
                        type="filepath"
                    )
                    upload_btn = gr.Button("Upload Documents", variant="primary")
                    upload_result = gr.Textbox(label="Upload Results", lines=6, interactive=False)
            
            with gr.Row():
                status_box2 = gr.Textbox(value=get_status(), label="System Status", lines=15, interactive=False)
            
            upload_btn.click(
                fn=add_documents,
                inputs=file_upload,
                outputs=[upload_result, status_box2]
            )
        
        # TAB 3: Add Web Pages
        with gr.TabItem("🌐 Add Web Pages"):
            gr.Markdown("### Add Multiple Web Pages")
            gr.Markdown("Enter URLs (one per line)")
            
            with gr.Row():
                with gr.Column():
                    web_urls = gr.Textbox(
                        label="Web URLs",
                        placeholder="https://example1.com\nhttps://example2.com",
                        lines=5
                    )
                    web_btn = gr.Button("Add Web Pages", variant="primary")
                    web_result = gr.Textbox(label="Web Results", lines=6, interactive=False)
            
            with gr.Row():
                status_box3 = gr.Textbox(value=get_status(), label="System Status", lines=15, interactive=False)
            
            web_btn.click(
                fn=add_web_urls,
                inputs=web_urls,
                outputs=[web_result, status_box3]
            )
        
        # TAB 4: Tickets
        with gr.TabItem("🎫 Support Tickets"):
            gr.Markdown("### Support Tickets")
            gr.Markdown("Tickets are created when the AI cannot answer with confidence.")
            
            def get_tickets():
                tickets = load_tickets()
                tenant_tickets = [t for t in tickets if t.get("organization_id") == DEFAULT_ORGANIZATION_ID]
                if not tenant_tickets:
                    return "No tickets found."
                
                lines = []
                for t in tenant_tickets:
                    lines.append(f"""--- TICKET ---
ID: {t.get('ticket_id')}
Status: {t.get('status')}
Priority: {t.get('priority')}
Confidence: {t.get('confidence')}
Reason: {t.get('reason')}
Question: {t.get('query')}
Created: {t.get('created_at')}
""")
                return "\n".join(lines)
            
            tickets_box = gr.Textbox(label="Tickets", lines=20, interactive=False)
            refresh_tickets = gr.Button("Refresh Tickets")
            refresh_tickets.click(fn=get_tickets, outputs=tickets_box)
            
            # Load tickets on startup
            tickets_box.value = get_tickets()
        
        # TAB 5: Rebuild
        with gr.TabItem("🔄 Rebuild"):
            gr.Markdown("### Rebuild Knowledge Base")
            gr.Markdown("Rebuild chunks and embeddings from all documents.")
            
            def do_rebuild():
                rebuild_knowledge_base(DEFAULT_ORGANIZATION_ID)
                return "Knowledge base rebuilt successfully!", get_status()
            
            rebuild_btn = gr.Button("Rebuild Knowledge Base", variant="primary")
            rebuild_result = gr.Textbox(label="Rebuild Result", lines=5, interactive=False)
            rebuild_status = gr.Textbox(value=get_status(), label="Status", lines=15, interactive=False)
            
            rebuild_btn.click(
                fn=do_rebuild,
                outputs=[rebuild_result, rebuild_status]
            )

# ------------------------------------------------------------
# LAUNCH
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("SUPPORTIQ WEB SYSTEM READY")
print("=" * 60)
print(f"Organization: {DEFAULT_ORGANIZATION_ID}")
print("\nFeatures:")
print("  📄 Upload multiple documents (PDF, DOCX, TXT)")
print("  🌐 Add multiple web pages")
print("  💬 Ask questions with citations")
print("  🎫 View support tickets")
print("  🔄 Rebuild knowledge base")
print("=" * 60)

demo.queue()
demo.launch(share=True, debug=False)

print("\n" + "=" * 60)
print("CELL 04 COMPLETE")
print("=" * 60)
```

```python
# ============================================================
# SUPPORTIQ — CELL 04
# TERMINAL COMPLETE SYSTEM
# ============================================================

import os
import time
from pathlib import Path

print("=" * 60)
print("SUPPORTIQ - TERMINAL COMPLETE SYSTEM")
print("=" * 60)

# ------------------------------------------------------------
# MAIN MENU
# ------------------------------------------------------------

def show_menu():
    print("\n" + "=" * 60)
    print("SUPPORTIQ MAIN MENU")
    print("=" * 60)
    print("1. Upload Documents (PDF, DOCX, TXT)")
    print("2. Add Web Pages (URLs)")
    print("3. Ask Questions")
    print("4. Show Status")
    print("5. View Tickets")
    print("6. Rebuild Knowledge Base")
    print("7. Clear All Data")
    print("8. Exit")
    print("=" * 60)

def upload_documents():
    print("\n" + "-" * 40)
    print("UPLOAD DOCUMENTS")
    print("-" * 40)
    print("Enter file paths (one per line, empty line to finish)")
    print("Supported: PDF, DOCX, TXT")
    print("Example: /kaggle/input/myfile.pdf")
    print()
    
    files = []
    while True:
        path = input("File path: ").strip()
        if not path:
            break
        if os.path.exists(path):
            files.append(path)
            print(f"  ✓ Added: {os.path.basename(path)}")
        else:
            print(f"  ✗ File not found: {path}")
    
    if not files:
        print("No files added.")
        return
    
    print(f"\nUploading {len(files)} files...")
    results = []
    for file_path in files:
        try:
            result = ingest_local_file(file_path, DEFAULT_ORGANIZATION_ID)
            if isinstance(result, dict):
                results.append(f"✓ {result.get('source_name')} - {result.get('document_id')}")
            else:
                results.append(f"✗ {os.path.basename(file_path)} - {result}")
        except Exception as e:
            results.append(f"✗ {os.path.basename(file_path)} - {str(e)}")
    
    print("\nResults:")
    for r in results:
        print(f"  {r}")
    
    # Rebuild index
    print("\nRebuilding knowledge base...")
    rebuild_knowledge_base(DEFAULT_ORGANIZATION_ID)
    print("Done!")

def add_web_pages():
    print("\n" + "-" * 40)
    print("ADD WEB PAGES")
    print("-" * 40)
    print("Enter URLs (one per line, empty line to finish)")
    print("Example: https://example.com")
    print()
    
    urls = []
    while True:
        url = input("URL: ").strip()
        if not url:
            break
        urls.append(url)
        print(f"  ✓ Added: {url}")
    
    if not urls:
        print("No URLs added.")
        return
    
    print(f"\nAdding {len(urls)} web pages...")
    results = []
    for url in urls:
        try:
            result = ingest_web_url(url, DEFAULT_ORGANIZATION_ID)
            if isinstance(result, dict):
                results.append(f"✓ {url} - {result.get('document_id')}")
            else:
                results.append(f"✗ {url} - {result}")
        except Exception as e:
            results.append(f"✗ {url} - {str(e)}")
    
    print("\nResults:")
    for r in results:
        print(f"  {r}")
    
    # Rebuild index
    print("\nRebuilding knowledge base...")
    rebuild_knowledge_base(DEFAULT_ORGANIZATION_ID)
    print("Done!")

def ask_questions():
    print("\n" + "-" * 40)
    print("ASK QUESTIONS")
    print("-" * 40)
    print("Ask anything about your uploaded documents.")
    print("Type 'back' to return to main menu.")
    print()
    
    # Check if any documents exist
    docs = load_documents()
    tenant_docs = [d for d in docs if d.get("organization_id") == DEFAULT_ORGANIZATION_ID]
    if not tenant_docs:
        print("⚠️ No documents found! Please upload documents first.")
        return
    
    # Check if index exists
    embeddings, _ = load_current_index()
    if len(embeddings) == 0:
        print("⚠️ No index found! Rebuilding...")
        rebuild_knowledge_base(DEFAULT_ORGANIZATION_ID)
    
    while True:
        print("\n" + "-" * 40)
        query = input("Question (or 'back'): ").strip()
        
        if query.lower() in ['back', 'b']:
            break
        
        if not query:
            print("Please enter a question.")
            continue
        
        print("\nProcessing...")
        start = time.time()
        result = supportiq_answer(query, DEFAULT_ORGANIZATION_ID)
        elapsed = time.time() - start
        
        print("\n" + "=" * 60)
        print("ANSWER:")
        print("-" * 60)
        print(result.get("answer", "No answer."))
        
        print("\nCITATIONS:")
        print("-" * 60)
        citations = result.get("citations", [])
        if citations:
            for c in citations:
                print(f"  • {c}")
        else:
            print("  No citations.")
        
        print("\nSTATUS:")
        print("-" * 60)
        print(f"  Status: {result.get('status', 'UNKNOWN')}")
        print(f"  Confidence: {result.get('confidence', 0.0):.1%}")
        print(f"  Latency: {elapsed:.2f}s")
        
        if result.get("ticket_id"):
            print(f"  Ticket: {result.get('ticket_id')}")
        print("=" * 60)

def show_status():
    docs = load_documents()
    tenant_docs = [d for d in docs if d.get("organization_id") == DEFAULT_ORGANIZATION_ID]
    chunks = load_chunks(DEFAULT_ORGANIZATION_ID)
    embeddings, _ = load_current_index()
    tickets = load_tickets()
    tenant_tickets = [t for t in tickets if t.get("organization_id") == DEFAULT_ORGANIZATION_ID]
    
    print("\n" + "=" * 60)
    print("SUPPORTIQ STATUS")
    print("=" * 60)
    print(f"Organization: {DEFAULT_ORGANIZATION_ID}")
    print("")
    print(f"📄 Documents: {len(tenant_docs)}")
    if tenant_docs:
        for d in tenant_docs:
            print(f"   - {d.get('source_name')} ({d.get('character_count', 0)} chars)")
    print("")
    print(f"📦 Chunks: {len(chunks)}")
    print(f"🔢 Embeddings: {len(embeddings)}")
    print(f"🎫 Tickets: {len(tenant_tickets)}")
    if tenant_tickets:
        for t in tenant_tickets:
            print(f"   - {t.get('ticket_id')} | {t.get('status')} | {t.get('query')[:30]}...")
    print("")
    print(f"🟢 Status: {'READY' if len(embeddings) > 0 else 'NO INDEX'}")
    print("=" * 60)

def view_tickets():
    tickets = load_tickets()
    tenant_tickets = [t for t in tickets if t.get("organization_id") == DEFAULT_ORGANIZATION_ID]
    
    print("\n" + "=" * 60)
    print("SUPPORT TICKETS")
    print("=" * 60)
    
    if not tenant_tickets:
        print("No tickets found.")
        return
    
    for i, t in enumerate(tenant_tickets, 1):
        print(f"\n--- TICKET #{i} ---")
        print(f"ID: {t.get('ticket_id')}")
        print(f"Status: {t.get('status')}")
        print(f"Priority: {t.get('priority')}")
        print(f"Confidence: {t.get('confidence')}")
        print(f"Reason: {t.get('reason')}")
        print(f"Question: {t.get('query')}")
        print(f"AI Answer: {t.get('ai_answer', '')[:100]}...")
        print(f"Created: {t.get('created_at')}")
    
    print("\n" + "=" * 60)

def rebuild_kb():
    print("\n" + "-" * 40)
    print("REBUILD KNOWLEDGE BASE")
    print("-" * 40)
    
    docs = load_documents()
    tenant_docs = [d for d in docs if d.get("organization_id") == DEFAULT_ORGANIZATION_ID]
    
    if not tenant_docs:
        print("No documents found. Please upload documents first.")
        return
    
    confirm = input(f"Rebuild index from {len(tenant_docs)} documents? (y/n): ").strip().lower()
    if confirm != 'y':
        print("Cancelled.")
        return
    
    print("\nRebuilding...")
    start = time.time()
    rebuild_knowledge_base(DEFAULT_ORGANIZATION_ID)
    elapsed = time.time() - start
    
    print(f"✅ Done! ({elapsed:.2f}s)")

def clear_all():
    print("\n" + "-" * 40)
    print("CLEAR ALL DATA")
    print("-" * 40)
    print("⚠️  This will delete all documents, chunks, embeddings, and tickets!")
    
    confirm = input("Are you sure? (yes/no): ").strip().lower()
    if confirm != 'yes':
        print("Cancelled.")
        return
    
    confirm2 = input("Type 'DELETE' to confirm: ").strip()
    if confirm2 != 'DELETE':
        print("Cancelled.")
        return
    
    # Delete files
    files_to_delete = [
        DOCUMENTS_JSON, CHUNKS_JSON, 
        EMBEDDINGS_PATH, METADATA_PATH, TICKETS_JSON
    ]
    
    for f in files_to_delete:
        if f.exists():
            f.unlink()
            print(f"  Deleted: {f.name}")
    
    print("\n✅ All data cleared!")

# ------------------------------------------------------------
# MAIN LOOP
# ------------------------------------------------------------

def main():
    print("\n" + "=" * 60)
    print("SUPPORTIQ TERMINAL SYSTEM")
    print("=" * 60)
    print(f"Organization: {DEFAULT_ORGANIZATION_ID}")
    
    # Check if documents exist
    docs = load_documents()
    tenant_docs = [d for d in docs if d.get("organization_id") == DEFAULT_ORGANIZATION_ID]
    if tenant_docs:
        print(f"📄 Found {len(tenant_docs)} documents")
        chunks = load_chunks(DEFAULT_ORGANIZATION_ID)
        print(f"📦 Found {len(chunks)} chunks")
        
        # Check if index exists
        embeddings, _ = load_current_index()
        if len(embeddings) == 0:
            print("⚠️  No index found. Rebuilding...")
            rebuild_knowledge_base(DEFAULT_ORGANIZATION_ID)
    
    while True:
        show_menu()
        choice = input("\nEnter your choice (1-8): ").strip()
        
        if choice == '1':
            upload_documents()
        elif choice == '2':
            add_web_pages()
        elif choice == '3':
            ask_questions()
        elif choice == '4':
            show_status()
        elif choice == '5':
            view_tickets()
        elif choice == '6':
            rebuild_kb()
        elif choice == '7':
            clear_all()
        elif choice == '8':
            print("\nGoodbye!")
            break
        else:
            print("Invalid choice. Please enter 1-8.")
        
        input("\nPress Enter to continue...")

# ------------------------------------------------------------
# RUN
# ------------------------------------------------------------

if __name__ == "__main__":
    main()

print("\n" + "=" * 60)
print("CELL 04 COMPLETE")
print("=" * 60)
```
